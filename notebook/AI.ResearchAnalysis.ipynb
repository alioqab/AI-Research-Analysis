## NLP 1 - Text Cleaning

```python
df["abstract"].iloc[0]
```

```python
stop_words = set(stopwords.words("english"))
```

```python
def clean_text(text):

    # lowercase
    text = text.lower()

    # remove punctuation
    text = re.sub(r"[^\w\s]", " ", text)

    # remove numbers
    text = re.sub(r"\d+", "", text)

    # tokenize
    words = word_tokenize(text)

    # remove stopwords
    words = [
        word
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)
```

```python
df["clean_abstract"] = (
    df["abstract"]
    .apply(clean_text)
)
```

```python
print("ORIGINAL:\n")
print(df["abstract"].iloc[0])

print("\n" + "="*80 + "\n")

print("CLEANED:\n")
print(df["clean_abstract"].iloc[0])
```

```python


vectorizer = CountVectorizer()

X = vectorizer.fit_transform(
    df["clean_abstract"]
)

word_counts = X.sum(axis=0)

word_freq = pd.DataFrame({
    "word": vectorizer.get_feature_names_out(),
    "count": word_counts.A1
})

word_freq = (
    word_freq
    .sort_values(
        by="count",
        ascending=False
    )
)

word_freq.head(30)
```

```python
research_stopwords = {
    "paper",
    "papers",
    "method",
    "methods",
    "approach",
    "approaches",
    "work",
    "study",
    "propose",
    "proposed",
    "show",
    "results",
    "result",
    "demonstrate",
    "first",
    "however",
    "existing",
    "introduce",
    "introduced",
    "using",
    "based",
    "github",
    "https",
    "com",
    "available",
    "framework",
    "performance",
    "experiments",
    "experiment",
    "state",
    "art",
    "task",
    "tasks",
    "system",
    "systems"
}

all_stopwords = stop_words.union(
    research_stopwords
)
```

Clean text V2

```python
def clean_text_v2(text):

    text = text.lower()

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\d+", "", text)

    words = word_tokenize(text)

    words = [
        word
        for word in words
        if word not in all_stopwords
    ]

    return " ".join(words)
```

```python
df["clean_abstract_v2"] = (
    df["abstract"]
    .apply(clean_text_v2)
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(
    df["clean_abstract_v2"]
)

word_counts = X.sum(axis=0)

word_freq = pd.DataFrame({
    "word": vectorizer.get_feature_names_out(),
    "count": word_counts.A1
})

word_freq = (
    word_freq
    .sort_values(
        by="count",
        ascending=False
    )
)

word_freq.head(30)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

def get_top_ngrams(text_series,
                   ngram_range=(3,3),
                   top_n=20):

    vectorizer = CountVectorizer(
        ngram_range=ngram_range
    )

    X = vectorizer.fit_transform(
        text_series
    )

    counts = X.sum(axis=0)

    ngrams = pd.DataFrame({
        "phrase": vectorizer.get_feature_names_out(),
        "count": counts.A1
    })

    ngrams = (
        ngrams
        .sort_values(
            by="count",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )

    return ngrams
```

```python
cv_trigrams = get_top_ngrams(
    cv_df["clean_abstract_v2"],
    ngram_range=(3,3)
)

cv_trigrams
```

```python
lg_trigrams = get_top_ngrams(
    lg_df["clean_abstract_v2"],
    ngram_range=(3,3)
)

lg_trigrams
```
